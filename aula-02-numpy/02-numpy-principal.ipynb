## 1. 🔁 Revisão de listas

Você já conhece a **lista** do Python: ela guarda uma sequência de valores entre colchetes. Vamos usar uma lista com as **vendas de uma semana** de uma loja, em reais, de segunda a domingo.

```python
# Esta célula cria a lista com as vendas da semana.
vendas_semana = [1200, 950, 1430, 780, 1650, 2100, 1800]
vendas_semana
```

### 1.1 O que a lista faz bem

Pegar um valor pela posição (a contagem começa no **0**), pegar um pedaço (**fatia**) e usar funções como `sum`, `max` e `len`.

```python
# Esta célula mostra a venda de segunda (primeira posição) e a de domingo (última).
print("segunda:", vendas_semana[0])
print("domingo:", vendas_semana[-1])
```

```python
# Esta célula pega os 3 primeiros dias da semana (segunda a quarta).
vendas_semana[:3]
```

```python
# Esta célula calcula o total, o maior valor e a média da semana.
total = sum(vendas_semana)
media = total / len(vendas_semana)

print("total:", total)
print("maior venda:", max(vendas_semana))
print("média:", round(media, 2))
```

Total de **R$ 9.910**, maior venda de **R$ 2.100** e média de cerca de **R$ 1.416** por dia. Até aqui a lista se saiu bem.

### 1.2 Aplicando uma conta em todos os valores

Pediram um **desconto de 10%** em todas as vendas (uma simulação). Com lista, precisamos de um laço `for`:

```python
# Esta célula aplica o desconto de 10% em cada venda, usando um for.
vendas_com_desconto = []
for venda in vendas_semana:
    vendas_com_desconto.append(venda * 0.9)

vendas_com_desconto
```

Funcionou, mas foram **3 linhas** para uma conta simples. E se pedirem para contar os dias acima de R$ 1.500? Outro laço:

```python
# Esta célula conta quantos dias tiveram venda acima de R$ 1.500.
dias_acima = 0
for venda in vendas_semana:
    if venda > 1500:
        dias_acima += 1

dias_acima
```

Foram **3 dias** (quinta, sábado e domingo). Cada pergunta nova pede um laço novo.

### 1.3 Onde a lista trava

A gente tenta o jeito "natural": multiplicar a lista por um número.

```python
# Esta célula multiplica a lista por 2, esperando dobrar as vendas.
vendas_semana * 2
```

Em vez de dobrar cada venda, a lista **se repetiu** (14 valores). Para a lista, `*` significa "repita".

Agora o desconto de 10% do jeito natural. **Vou provocar o erro de propósito:**

```python
# Esta célula tenta multiplicar a lista por 0.9 (vai dar erro).
vendas_semana * 0.9
```

Como ler a mensagem: `TypeError: can't multiply sequence by non-int of type 'float'`. Traduzindo: "não consigo multiplicar uma **sequência** (a lista) por um número que não é inteiro". A lista só entende `* 2` como repetição.

> 💡 **A lista guarda qualquer coisa** (números, textos, outras listas). Justamente por isso ela **não sabe fazer conta** com os valores. Para cálculo com números, existe uma ferramenta feita sob medida: o **array do NumPy**.

## 2. 🧮 O que é um vetor

Um **vetor** (ou **array** de uma dimensão) é como uma lista, mas com duas regras:

- só guarda **números do mesmo tipo**;
- já vem com as **contas prontas**: soma, média, desconto, comparação, tudo de uma vez.

### 2.1 Criando um array

O jeito mais simples é converter uma lista com `np.array(...)`.

```python
# Esta célula converte a lista de vendas em um array do NumPy.
vendas = np.array(vendas_semana)
vendas
```

```python
# Esta célula confere o tipo do objeto criado.
type(vendas)
```

O nome técnico é **`ndarray`** (*n-dimensional array*). Pode chamar de array. Agora repare nas mesmas contas que travaram na lista:

```python
# Esta célula dobra todas as vendas.
vendas * 2
```

```python
# Esta célula aplica o desconto de 10% em todas as vendas.
vendas * 0.9
```

```python
# Esta célula soma R$ 100 a todas as vendas.
vendas + 100
```

Sem laço e sem erro: **uma linha**, e o NumPy aplica a conta em cada valor. Isso se chama **vetorização**.

E a pergunta "quais dias passaram de R$ 1.500?" também vira uma linha:

```python
# Esta célula compara cada venda com 1500 e devolve verdadeiro ou falso para cada dia.
vendas > 1500
```

O resultado é um array de `True`/`False`, um para cada dia. Mais adiante você vai ver que ele serve de **filtro**.

### Outras formas de criar arrays

Nem sempre temos uma lista pronta. O NumPy cria sequências e arrays "em branco":

```python
# Esta célula cria os números de 1 a 7 (o 8 não entra).
dias = np.arange(1, 8)
dias
```

```python
# Esta célula cria um array de 5 zeros, útil para começar um acumulador.
np.zeros(5)
```

```python
# Esta célula cria 5 valores igualmente espaçados entre 0 e 1.
np.linspace(0, 1, 5)
```

- `arange(início, fim)`: sequência de inteiros, **sem incluir o fim**;
- `zeros(n)`: `n` zeros;
- `linspace(início, fim, n)`: `n` valores igualmente espaçados, **incluindo o fim**.

Para praticar, vamos **gerar** os dados em vez de baixar arquivos. O NumPy tem um gerador de números aleatórios. Fixamos uma **semente** (`42`) para que **todos os alunos vejam exatamente os mesmos números**.

```python
# Esta célula cria o gerador de números aleatórios com a semente 42.
rng = np.random.default_rng(42)
```

### 2.2 Conhecendo o NumPy

Antes de resolver os casos, uma **visita guiada** pelas ideias que vamos usar. Cada uma aparece aqui em pequeno, com poucos números, para você enxergar o que acontece; depois aplicamos nos casos.

#### Olhando o array

Passo 2, olhar o array: `shape` (formato), `size` (quantidade de valores) e `dtype` (tipo dos números).

```python
# Esta célula mostra o formato, a quantidade de valores e o tipo do array.
print("shape:", vendas_30_dias.shape)
print("size :", vendas_30_dias.size)
print("dtype:", vendas_30_dias.dtype)
```

O `shape` `(30,)` quer dizer "uma única dimensão com 30 valores" (um vetor). O `dtype` `int64` indica números inteiros.

#### Estatísticas

Passo 4, as respostas para o gestor. Cada uma é uma linha:

```python
# Esta célula calcula as principais estatísticas das vendas do mês.
print("total do mês :", vendas_30_dias.sum())
print("média diária :", vendas_30_dias.mean().round(2))
print("menor venda  :", vendas_30_dias.min())
print("maior venda  :", vendas_30_dias.max())
```

`argmax` devolve a **posição** do maior valor, e não o valor em si. Aqui deu `11`: como a contagem começa no 0, a maior venda (R$ 2.939) foi no **12º dia**.

> 💡 **Compare com a lista:** para cada uma dessas perguntas, a lista pediria um laço ou uma função separada. No NumPy, o próprio array "sabe" fazer a conta.

#### Pegando partes: posição e fatias

Cada valor do array tem uma **posição**, e a contagem começa no **0**. Posição negativa conta de trás para frente (`-1` é o último). Uma **fatia** `[início:fim:passo]` pega um pedaço, e o **fim não entra**. É igual ao que você já fez com listas.

```python
# Esta célula mostra posição e fatias do array de vendas da semana.
print("primeira:", vendas[0])
print("última  :", vendas[-1])
print("3 primeiras:", vendas[:3])
print("2 últimas  :", vendas[-2:])
print("de 2 em 2  :", vendas[::2])
```

Se pedirmos uma posição que não existe, o NumPy dá `IndexError`. E há uma **pegadinha**: uma fatia não é uma cópia do array, é uma "janela" para os mesmos dados. Vamos ver isso funcionando no caso financeiro.

#### Máscaras booleanas: o filtro do NumPy

Lá no começo, `vendas > 1500` devolveu um array de `True`/`False`. Esse array se chama **máscara**: ele diz, para cada posição, se o valor **atende à condição**. Usada entre colchetes, a máscara vira um **filtro**: sobram só os valores com `True`.

```python
# Esta célula guarda a máscara dos dias com venda acima de 1500 e usa como filtro.
mascara = vendas > 1500

print("máscara :", mascara)
print("filtrado:", vendas[mascara])
```

O filtro devolveu só as três vendas acima de R$ 1.500. Como `True` vale 1 e `False` vale 0, a máscara também **conta** e ajuda a **calcular**:

```python
# Esta célula conta os dias acima de 1500 e calcula a média só desses dias.
print("quantos dias     :", mascara.sum())
print("média desses dias:", vendas[mascara].mean())
```

Para combinar condições, usamos três operadores. **Cada condição vai entre parênteses.**

| Operador | Significa | Exemplo |
|---|---|---|
| `&` | **e** (as duas condições) | `(vendas >= 1000) & (vendas <= 1700)` |
| `\|` | **ou** (pelo menos uma) | `(vendas < 900) \| (vendas > 2000)` |
| `~` | **não** (inverte a máscara) | `~(vendas > 1500)` |

```python
# Esta célula testa os três operadores com as vendas da semana.
print("entre 1000 e 1700             :", (vendas >= 1000) & (vendas <= 1700))
print("abaixo de 900 ou acima de 2000:", (vendas < 900) | (vendas > 2000))
print("NÃO acima de 1500             :", ~(vendas > 1500))
```

Dá também para **alterar só quem atende à condição**. Aqui, trabalhando numa cópia, subimos para R$ 1.000 as vendas abaixo disso:

```python
# Esta célula cria uma cópia e substitui as vendas abaixo de 1000 por 1000.
ajustadas = vendas.copy()
ajustadas[ajustadas < 1000] = 1000
ajustadas
```

Os dois valores abaixo de R$ 1.000 (950 e 780) viraram 1000, e o array `vendas` original não mudou.

#### Tabelas (matrizes) e o `axis`

Um array também pode ter **linhas e colunas**. Aqui, 2 agências (linhas) e 3 meses (colunas):

```python
# Esta célula cria uma tabela com 2 agências (linhas) e 3 meses (colunas).
tabela = np.array([[100, 120, 90],
                   [80, 95, 110]])

print(tabela)
print("shape:", tabela.shape)
```

O `shape` `(2, 3)` é sempre **(linhas, colunas)**. Para pegar partes, usamos `[linha, coluna]`, e o `:` sozinho significa "todos".

```python
# Esta célula pega uma linha, uma coluna e um único valor da tabela.
print("agência 0 (linha 0):", tabela[0])
print("mês 0 (coluna 0)   :", tabela[:, 0])
print("agência 1, mês 2   :", tabela[1, 2])
```

Funções como `sum` e `mean` aceitam o **`axis`**, que diz **em que direção** fazer a conta:

- **`axis=0`**: desce pelas **linhas** e dá **uma resposta por coluna** (aqui, um total por mês);
- **`axis=1`**: anda pelas **colunas** e dá **uma resposta por linha** (aqui, um total por agência).

```python
# Esta célula soma a tabela por coluna (axis=0) e por linha (axis=1).
print("total por mês     (axis=0):", tabela.sum(axis=0))
print("total por agência (axis=1):", tabela.sum(axis=1))
```

Conferindo na mão: o mês 0 teve 100 + 80 = **180** e a agência 0 vendeu 100 + 120 + 90 = **310**. Sem `axis`, o NumPy soma tudo e devolve um único número.

#### Broadcasting: arrays de tamanhos diferentes

E se quisermos multiplicar **cada mês por um fator diferente**? Temos a tabela `(2, 3)` e um vetor com 3 fatores, um por coluna. O NumPy "estica" o vetor para todas as linhas e faz a conta. Isso se chama **broadcasting**.

```python
# Esta célula multiplica cada coluna da tabela pelo seu fator.
fatores = np.array([1.0, 1.1, 1.2])
tabela * fatores
```

Cada coluna foi multiplicada pelo seu fator: o mês 0 ficou igual (×1,0), o mês 1 subiu 10% e o mês 2 subiu 20%. Isso só funciona quando os formatos se **encaixam**; se não, vem um `ValueError`. Vamos ver um caso desses mais adiante.

#### Valores faltantes: `np.nan`

Em dados reais, é comum faltar informação. No NumPy, um valor faltante é o **`np.nan`** ("not a number"). Ele **contamina** as contas, e por isso existem funções que o ignoram, com o prefixo `nan`:

```python
# Esta célula mostra o efeito do nan na média e como ignorá-lo.
com_falta = np.array([100.0, np.nan, 300.0])

print("média comum   :", com_falta.mean())
print("média sem nan :", np.nanmean(com_falta))
print("quem é nan    :", np.isnan(com_falta))
```

A média comum virou `nan`, e `np.nanmean` ignorou o faltante e calculou (100 + 300) / 2 = **200**. Para **achar** os faltantes use `np.isnan`; nunca `== np.nan`, porque um `nan` nunca é igual a nada. Vamos ver isso com mais calma no caso de RH.

### 2.3 Comparação de velocidade

Com 7 vendas ninguém percebe diferença. Mas uma base de clientes pode ter **milhões de linhas**. Vamos medir: aplicar o desconto de 10% em **1 milhão de valores**, com lista e com array. O comando `%timeit` repete a conta várias vezes e mostra o tempo médio.

```python
# Esta célula cria 1 milhão de valores, uma vez como lista e outra como array.
um_milhao_lista = list(range(1_000_000))
um_milhao_array = np.arange(1_000_000)
```

```python
# Esta célula mede o tempo do desconto feito com lista e for (list comprehension).
%timeit [valor * 0.9 for valor in um_milhao_lista]
```

```python
# Esta célula mede o tempo do desconto feito com array.
%timeit um_milhao_array * 0.9
```

No meu computador, a lista levou cerca de **50 milissegundos** e o array cerca de **2**: o array foi umas **25 vezes mais rápido**, fazendo exatamente a mesma conta.

Por que o array ganha?
- guarda **só um tipo de número**, lado a lado na memória;
- o laço roda em **código compilado (C)**, e não no `for` do Python.

> 💡 Os tempos variam de computador para computador, mas a diferença de ordem de grandeza se mantém.

## 4.1 💰 Caso financeiro: saldos de clientes

### De onde vêm os dados?

São dados **sintéticos**, gerados na hora com a semente `42`. Parecem uma carteira de clientes, mas **não são clientes de verdade**.

| Variável | O que é | Unidade |
|---|---|---|
| `saldos` | Saldo em conta corrente de 500 clientes. **Negativo** = cliente no cheque especial | R$ |
| `anos_de_conta` | Há quantos anos cada cliente tem conta (de 1 a 20), na mesma ordem de `saldos` | anos |

```python
# Esta célula gera os 500 saldos e o tempo de conta de cada cliente.
rng = np.random.default_rng(42)
saldos = rng.normal(loc=5000, scale=3000, size=500).round(2)
anos_de_conta = rng.integers(1, 21, size=500)
```

```python
# Esta célula cria uma versão em lista dos saldos, para resolvermos também com lista.
saldos_lista = saldos.tolist()
```

### Conhecendo os dados: posição e fatias

Lembra de posição e fatia? Agora com os 500 clientes.

```python
# Esta célula mostra o primeiro e o último cliente.
print("primeiro:", saldos[0])
print("último  :", saldos[-1])
```

```python
# Esta célula mostra os 5 primeiros saldos.
saldos[:5]
```

```python
# Esta célula mostra um cliente a cada 100 (passo de 100).
saldos[::100]
```

O índice 0 é o primeiro cliente e o `-1` é o último. A última fatia pegou os clientes 0, 100, 200, 300 e 400.

O que acontece se pedirmos uma posição que não existe? **Vou provocar o erro de propósito:**

```python
# Esta célula pede o cliente de posição 500 (só existem as posições de 0 a 499).
saldos[500]
```

Como ler: `IndexError: index 500 is out of bounds for axis 0 with size 500`. Em português: "o índice 500 está fora dos limites; o eixo tem tamanho 500". Com 500 clientes, as posições válidas vão de **0 a 499**.

### Pergunta 1: quantos clientes estão no negativo?

**Com lista:** um laço que separa os saldos negativos.

```python
# Esta célula separa os saldos negativos usando lista e for.
negativos_lista = []
for saldo in saldos_lista:
    if saldo < 0:
        negativos_lista.append(saldo)

len(negativos_lista)
```

**Com NumPy:** comparamos o array inteiro com zero. O resultado é uma **máscara**, um array de `True`/`False` com um valor para cada cliente.

```python
# Esta célula cria a máscara de clientes no negativo.
mascara_negativo = saldos < 0
mascara_negativo[:8]
```

```python
# Esta célula conta os True da máscara (True vale 1, False vale 0).
mascara_negativo.sum()
```

Mesma resposta (**22 clientes**), sem laço. E a máscara também funciona como **filtro**: colocada entre colchetes, ela devolve só os valores onde está `True`.

```python
# Esta célula usa a máscara como filtro e mostra os 8 primeiros saldos negativos.
saldos[mascara_negativo][:8]
```

### Pergunta 2: qual o saldo médio de quem está no negativo?

**Com lista:**

### Pergunta 4: quantos clientes têm saldo alto E conta antiga?

Queremos saldo acima de R$ 8.000 **e** mais de 10 anos de conta. **Com lista**, precisamos percorrer as duas listas juntas:

```python
# Esta célula conta quem tem saldo > 8000 e 10 anos ou mais de conta, usando lista.
contagem = 0
for saldo, anos in zip(saldos_lista, anos_de_conta.tolist()):
    if saldo > 8000 and anos >= 10:
        contagem += 1

contagem
```

**Com NumPy**, combinamos duas máscaras com `&` (e). Cada condição vai **entre parênteses**.

```python
# Esta célula combina duas condições com & e conta os clientes.
saldo_alto_e_antigo = (saldos > 8000) & (anos_de_conta >= 10)
saldo_alto_e_antigo.sum()
```

São **44 clientes** com saldo alto e conta antiga. Repare nos parênteses em volta de cada condição: eles são obrigatórios.

Lembra dos operadores `&` (e), `|` (ou) e `~` (não)? Existem **dois erros muito comuns** aqui. O primeiro é usar o `and` do Python no lugar do `&`. **Vou provocar o erro:**

```python
# Esta célula usa and em vez de & (vai dar erro).
(saldos > 8000) and (anos_de_conta >= 10)
```

Como ler: `ValueError: The truth value of an array with more than one element is ambiguous`. O `and` do Python espera **um** verdadeiro/falso, e recebeu 500. **Para arrays, use `&` e `|`.**

O segundo erro é **esquecer os parênteses**:

```python
# Esta célula esquece os parênteses nas comparações (vai dar erro).
saldos > 8000 & anos_de_conta >= 10
```

É o mesmo erro por outro motivo: o `&` é calculado **antes** das comparações. Regra de ouro: **cada comparação entre parênteses**.

### Pergunta 5: a pegadinha da fatia (cópia × referência)

Você pega os 3 primeiros valores para fazer uma simulação e zera um deles. Uma **fatia** parece um array novo, mas **não é**: ela é uma "janela" para os **mesmos dados**. Vamos testar com 5 contas.

```python
# Esta célula cria um array de teste com 5 contas e uma fatia com as 3 primeiras.
contas = np.array([1000, 2000, 3000, 4000, 5000])
fatia = contas[:3]
fatia
```

```python
# Esta célula zera o primeiro valor da fatia e mostra os dois arrays.
fatia[0] = 0
print("fatia :", fatia)
print("contas:", contas)
```

Mexemos só na `fatia`, mas **`contas` também mudou** (o 1000 virou 0). Os dois compartilham os mesmos dados. Em uma base de verdade, isso estragaria o original **sem avisar**.

> 💡 **Regra:** se vai alterar um pedaço e quer preservar o original, use **`.copy()`**.

```python
# Esta célula refaz o teste usando uma cópia da fatia.
contas = np.array([1000, 2000, 3000, 4000, 5000])
copia = contas[:3].copy()
copia[0] = 0

print("copia :", copia)
print("contas:", contas)
```

Agora `contas` ficou intacta.

### ✅ O que o caso financeiro mostrou

- 🎯 **Máscara** (`saldos < 0`) é o filtro do NumPy; `.sum()` conta e `.mean()` calcula a média do que foi filtrado.
- 🔗 Para combinar condições: **`&`, `|`, `~`**, sempre com parênteses.
- ⚠️ **Fatia não é cópia**: use `.copy()` quando precisar preservar o original.
- 🔢 Posição começa no 0; passar do limite dá `IndexError`.

## 4.2 🛒 Caso comercial: vendas por agência

### De onde vêm os dados?

Também **sintéticos**. Agora temos uma **tabela**: as vendas de produtos financeiros em **12 meses** (linhas) em **6 agências** (colunas).

| Variável | Formato | O que é | Unidade |
|---|---|---|---|
| `vendas_mensais` | `(12, 6)` | Vendas de cada agência (colunas) em cada mês (linhas). Linha 0 = janeiro; coluna 0 = agência 0 | R$ mil |
| `meta_mensal` | `(6,)` | Meta mensal de vendas de cada agência | R$ mil |

```python
# Esta célula gera a tabela de vendas (12 meses x 6 agências) e as metas.
rng_comercial = np.random.default_rng(7)
vendas_mensais = rng_comercial.integers(80, 200, size=(12, 6))
meta_mensal = np.array([120, 130, 110, 150, 125, 140])
```

```python
# Esta célula mostra o formato da tabela e os 3 primeiros meses.
print(vendas_mensais.shape)
vendas_mensais[:3]
```

O `shape` `(12, 6)` é sempre **(linhas, colunas)**: 12 meses e 6 agências. Para pegar partes de uma tabela, usamos `[linha, coluna]`, e o `:` sozinho significa "todos".

```python
# Esta célula mostra as vendas de janeiro (linha 0) em todas as agências.
vendas_mensais[0]
```

```python
# Esta célula mostra as vendas da agência 0 (coluna 0) nos 12 meses.
vendas_mensais[:, 0]
```

```python
# Esta célula mostra a venda da agência 2 em março (linha 2, coluna 2).
vendas_mensais[2, 2]
```

- `[0]`: uma **linha** inteira (um mês, todas as agências);
- `[:, 0]`: uma **coluna** inteira (todos os meses, uma agência);
- `[2, 2]`: **um único valor**.

### Pergunta 1: qual o total vendido por cada agência no ano?

**Com lista**, precisamos de uma lista de listas e de laços aninhados. Só para a agência 0:

```python
# Esta célula converte a tabela em lista de listas e soma a agência 0 com for.
vendas_lista = vendas_mensais.tolist()

total_agencia_0 = 0
for mes in vendas_lista:
    total_agencia_0 += mes[0]

total_agencia_0
```

E isso foi **só uma** agência. Para as 6, seriam laços dentro de laços. **Com NumPy**, a coluna 0 já resolve:

```python
# Esta célula soma a coluna da agência 0.
vendas_mensais[:, 0].sum()
```

Para somar **todas as agências de uma vez**, usamos o `axis`: **`axis=0`** dá uma resposta por **coluna** (6 valores, um por agência) e **`axis=1`** dá uma por **linha** (12 valores, um por mês).

```python
# Esta célula calcula o total do ano de cada agência.
total_por_agencia = vendas_mensais.sum(axis=0)
total_por_agencia
```

A **agência 3** vendeu mais no ano (R$ 1.848 mil) e a **agência 1** vendeu menos (R$ 1.503 mil). Por mês, o total vai de R$ 756 mil (abril) a R$ 1.019 mil (janeiro).

> 💡 **Dica para não confundir:** o eixo escolhido é o que **desaparece** do formato. Em `(12, 6)`, `axis=0` elimina o 12 e sobra `(6,)`; `axis=1` elimina o 6 e sobra `(12,)`.

### Pergunta 2: qual foi o melhor mês e a melhor agência?

```python
# Esta célula descobre o melhor mês (a contagem começa em 0, por isso somamos 1).
total_por_mes.argmax() + 1
```

O melhor mês foi o **1º (janeiro)** e a melhor agência foi a **agência 3**, lembrando que o índice das agências começa no 0.

### Pergunta 3: quem bateu a meta em cada mês?

Cada agência tem a sua meta. Comparamos a tabela `(12, 6)` com as metas `(6,)`, e o NumPy compara **cada coluna com a meta da sua agência**. É o broadcasting que vimos há pouco.

```python
# Esta célula compara cada venda com a meta da sua agência.
bateu_meta = vendas_mensais >= meta_mensal
bateu_meta[:3]
```

```python
# Esta célula conta, por agência, em quantos meses a meta foi batida.
bateu_meta.sum(axis=0)
```

A `agência 2` bateu a meta em **10 dos 12 meses**, e as agências 1 e 5 em apenas **6**. Como cada agência tem uma meta diferente, comparar só o total vendido não conta a história inteira.

Duas perguntas parecidas, usando `all` ("todos") e `any` ("algum"):

```python
# Esta célula verifica quais agências bateram a meta em TODOS os meses.
bateu_meta.all(axis=0)
```

```python
# Esta célula conta quantas agências bateram a meta em cada mês e mostra os 12 meses.
bateu_meta.sum(axis=1)
```

Agora o **erro de formato** mais comum. E se as metas fossem **por mês** (12 valores) em vez de por agência? **Vou provocar o erro:**

```python
# Esta célula tenta comparar a tabela (12, 6) com um vetor de 12 metas (vai dar erro).
meta_por_mes = np.full(12, 130)
vendas_mensais >= meta_por_mes
```

Como ler: `ValueError: operands could not be broadcast together with shapes (12,6) (12,)`. O NumPy compara os formatos **da direita para a esquerda**: `6` contra `12`. Não são iguais, então ele não consegue "esticar".

A correção é transformar o vetor em **coluna** `(12, 1)` com `[:, None]`. Assim o `12` bate com o `12`, e o `1` é esticado para 6.

```python
# Esta célula transforma as metas em coluna e refaz a comparação.
meta_por_mes_coluna = meta_por_mes[:, None]
print(meta_por_mes_coluna.shape)

(vendas_mensais >= meta_por_mes_coluna).shape
```

### ✅ O que o caso comercial mostrou

- 📋 Tabela: `shape` é **(linhas, colunas)**; `[linha, coluna]` e `:` para "todos".

- 📏 `argmax` dá a **posição** do maior valor.
- 🔄 **Broadcasting**: um vetor pode ser comparado com a tabela inteira, se os formatos se encaixarem. Se não, `ValueError`.

## 4.3 👥 Caso RH: reajuste salarial

### De onde vêm os dados?

Sintéticos, com a semente `3`.

| Variável | O que é | Unidade |
|---|---|---|
| `salarios` | Salário de 40 colaboradores, entre R$ 2.500 e R$ 11.999 | R$ |

**Regra do reajuste:** até R$ 4.000 → **8%**; de R$ 4.000 a R$ 8.000 → **5%**; acima de R$ 8.000 → **3%**.

```python
# Esta célula gera os salários de 40 colaboradores.
rng_rh = np.random.default_rng(3)
salarios = rng_rh.integers(2500, 12000, size=40)
salarios[:8]
```

### Pergunta 1: qual o novo salário de cada pessoa?

**Com lista**, um `if/elif/else` dentro do laço:

```python
# Esta célula calcula os novos salários com lista, for e if/elif/else.
novos_lista = []
for salario in salarios.tolist():
    if salario < 4000:
        novos_lista.append(salario * 1.08)
    elif salario <= 8000:
        novos_lista.append(salario * 1.05)
    else:
        novos_lista.append(salario * 1.03)

novos_lista[:5]
```

**Com NumPy**, começamos com um array de percentuais e preenchemos **só as posições escolhidas por uma máscara**. Isso é *atribuição por máscara*.

```python
# Esta célula cria o array de percentuais de reajuste, um para cada colaborador.
percentual = np.zeros(salarios.size)

percentual[salarios < 4000] = 0.08
percentual[(salarios >= 4000) & (salarios <= 8000)] = 0.05
percentual[salarios > 8000] = 0.03

percentual[:8]
```

```python
# Esta célula confere que ninguém ficou sem percentual (quantidade de zeros).
(percentual == 0).sum()
```

```python
# Esta célula calcula os novos salários.
novos_salarios = salarios * (1 + percentual)
novos_salarios[:5].round(2)
```

```python
# Esta célula confere se a versão com lista e a com NumPy dão o mesmo resultado.
np.allclose(novos_lista, novos_salarios)
```

`True`: as duas formas chegam ao mesmo resultado. Repare que o NumPy fez em **3 linhas de máscara** o que a lista fez com um laço e três `if`.

### Pergunta 2: qual o impacto do reajuste na folha?

```python
# Esta célula compara a folha de pagamento antes e depois do reajuste.
folha_antes = salarios.sum()
folha_depois = novos_salarios.sum()

print("folha antes :", folha_antes)
print("folha depois:", round(folha_depois, 2))
print("aumento     :", round(folha_depois - folha_antes, 2))
```

A folha subiu de **R$ 283.009** para **R$ 294.201,49**, um aumento de cerca de **R$ 11,2 mil** (aproximadamente 4%). Esse é o tipo de número que o gestor pede antes de aprovar um reajuste.

### Pergunta 3: e se faltarem salários na base?

Vamos simular 2 salários não informados, com o `np.nan` que vimos há pouco. Ele só existe em números **decimais**, por isso convertemos antes com `astype(float)`.

```python
# Esta célula cria uma cópia dos salários com 2 valores faltantes (posições 3 e 17).
salarios_incompletos = salarios.astype(float)
salarios_incompletos[[3, 17]] = np.nan
salarios_incompletos[:5]
```

```python
# Esta célula tenta calcular a média com os faltantes.
salarios_incompletos.mean()
```

Qualquer conta com um `nan` resulta em `nan`: ele "contamina" o resultado. As funções com prefixo **`nan`** ignoram os faltantes:

```python
# Esta célula calcula a média ignorando os valores faltantes.
np.nanmean(salarios_incompletos).round(2)
```

Para **achar** os faltantes, não use `==`. Um `nan` nunca é igual a nada, nem a ele mesmo:

```python
# Esta célula mostra que nan == nan é falso.
np.nan == np.nan
```

```python
# Esta célula acha os faltantes com np.isnan e conta quantos são.
faltantes = np.isnan(salarios_incompletos)
faltantes.sum()
```

```python
# Esta célula preenche os faltantes com a média dos salários conhecidos.
salarios_preenchidos = salarios_incompletos.copy()
salarios_preenchidos[faltantes] = np.nanmean(salarios_incompletos)
salarios_preenchidos[[3, 17]].round(2)
```

As duas posições agora têm a média dos salários conhecidos. Preencher com a média é uma decisão simples; em um projeto real, você escolheria com o time o melhor critério.

### ✅ O que o caso RH mostrou

- ✍️ **Atribuição por máscara**: `array[condição] = valor` altera só quem atende à condição.
- 🧮 Contas entre arrays do mesmo tamanho acontecem **posição por posição** (`salarios * (1 + percentual)`).
- ❓ **`np.nan`** = valor faltante: contamina as contas; use `np.nanmean`, `np.isnan`, e nunca `== np.nan`.

## 5. 🧱 Matriz, o essencial

Você já usou matrizes no caso comercial: a tabela `vendas_mensais` **é** uma matriz. Aqui vai um resumo de três ideias.

### 5.1 Reorganizar a tabela: `reshape`

`reshape` muda o formato **sem mudar os dados**. Os 12 meses da agência 0 podem virar **4 trimestres × 3 meses**:

```python
# Esta célula reorganiza as vendas da agência 0 em 4 trimestres de 3 meses.
por_trimestre = vendas_mensais[:, 0].reshape(4, 3)
por_trimestre
```

```python
# Esta célula soma cada linha, obtendo o total de cada trimestre.
por_trimestre.sum(axis=1)
```

Cada linha é um trimestre, e a soma por linha (`axis=1`) dá o total de cada um. O formato novo precisa ter **o mesmo total de elementos**. **Vou provocar o erro:**

```python
# Esta célula tenta reorganizar 12 valores em 5 linhas x 3 colunas (vai dar erro).
vendas_mensais[:, 0].reshape(5, 3)
```

Como ler: `ValueError: cannot reshape array of size 12 into shape (5,3)`. Quer dizer: "não dá para encaixar 12 valores em 5 × 3 = 15 posições". O produto das dimensões tem de dar o total.

### 5.2 Trocar linhas por colunas: transposta

`.T` gira a tabela: o que era linha vira coluna.

```python
# Esta célula mostra o formato da tabela e o da sua transposta.
print(vendas_mensais.shape, "->", vendas_mensais.T.shape)
```

### 5.3 Multiplicar matrizes: o `@`

`*` multiplica **posição por posição**. O **`@`** faz o **produto matricial**: multiplica cada linha pelas colunas e **soma**. É a conta de "quantidade × preço = receita".

Exemplo: 3 agências vendem 3 produtos. Queremos a **receita de cada agência**.

```python
# Esta célula cria as quantidades vendidas (agências x produtos) e o preço de cada produto.
quantidades = np.array([
    [10, 5, 2],   # agência A
    [ 8, 7, 3],   # agência B
    [12, 4, 1],   # agência C
])
precos = np.array([50.0, 30.0, 120.0])
```

```python
# Esta célula calcula a receita de cada agência com o produto matricial.
receita = quantidades @ precos
receita
```

Para a agência A: `10 × 50 + 5 × 30 + 2 × 120 = 890`. O `@` fez essa conta para as três agências de uma vez.

A regra para o `@` funcionar: o número de **colunas** do primeiro tem de ser igual ao número de **valores** do segundo (aqui, 3 produtos e 3 preços). **Vou provocar o erro:**

```python
# Esta célula usa só 2 preços para 3 produtos (vai dar erro).
quantidades @ np.array([50.0, 30.0])
```

Como ler: `ValueError: matmul: Input operand 1 has a mismatch in its core dimension 0 ... (size 2 is different from 3)`. Os 3 produtos não casam com 2 preços.

> 💡 **Não precisa decorar a matemática.** Guarde a ideia: `*` é posição por posição; `@` é "linha × coluna e soma", usado quando você tem quantidades e preços, ou pesos e valores.

| `np.arange`, `np.zeros`, `np.linspace` | cria sequências e arrays prontos |
| `rng = np.random.default_rng(42)` | gerador de dados de teste, sempre iguais |
| `shape`, `size`, `dtype` | formato, quantidade e tipo dos valores |
| `a[i]`, `a[i:j]`, `m[l, c]`, `m[:, c]` | posição, fatia, linha, coluna |
| `a.copy()` | cópia independente (fatia **não** é cópia) |
| `a > x`, `&`, `\|`, `~` | máscaras (filtros) e combinações |
| `a[mascara]`, `a[mascara] = valor` | filtrar / alterar só quem atende |
| `sum`, `mean`, `min`, `max`, `argmax` | estatísticas; aceitam `axis` |
| `axis=0` / `axis=1` | uma resposta por coluna / por linha |
| `np.nan`, `np.isnan`, `np.nanmean` | valores faltantes |
| `reshape`, `.T` | muda o formato / transpõe |
| `@` | produto matricial (quantidade × preço) |

**Erros que vimos e como ler:**

| Mensagem | Causa |
|---|---|
| `IndexError: index ... is out of bounds` | posição que não existe |
| `ValueError: The truth value of an array ... is ambiguous` | `and`/`or` no lugar de `&`/`\|`, ou falta de parênteses |
| `ValueError: operands could not be broadcast together` | formatos que não se encaixam |
| `ValueError: cannot reshape array of size ...` | o novo formato não tem o mesmo total de elementos |
| `ValueError: matmul: ... mismatch` | dimensões incompatíveis no `@` |
